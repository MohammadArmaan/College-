```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt 
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
```

```python
data = pd.read_csv("Iris.csv")
print(data)
```

```python
x = data.iloc[:,:-1].values
y = data.iloc[:,-1].values
```

```python
scaler = StandardScaler()
x_scaled = scaler.fit_transform(x)
```

```python
pca = PCA(n_components=2)
x_pca = pca.fit_transform(x_scaled)
```

```python
explained_variance = pca.explained_variance_ratio_
print("Explained Variance by each Principal Component")
print("Principal Component 1", explained_variance[0])
print("Principal Component 2", explained_variance[1])
print("Total Variance", sum(explained_variance))
```

```python
plt.figure(figsize=(10,6))
colors = ["red", "green", "blue"]
markers = ["o", "s", "D"]
labels = np.unique(y)

for i, color, marker, label in zip(range(len(labels)), colors, markers, labels):
    plt.scatter(x_pca[y == label, 0], x_pca[y == label, 1], alpha=0.7, color=color, label=label, marker=marker)
    
plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.legend()
plt.grid()
plt.show()
```
